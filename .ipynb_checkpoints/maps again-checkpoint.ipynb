# Additonal info

```python
# --- Unique station counts ---
n_start = df['start_station_name'].nunique()
n_end = df['end_station_name'].nunique()
all_stations = pd.concat([df['start_station_name'], df['end_station_name']])
n_total = all_stations.nunique()

print(f"Unique start stations: {n_start}")
print(f"Unique end stations:   {n_end}")
print(f"Total unique stations: {n_total}")

# --- Top 10 busiest stations (by total trips) ---
station_usage = all_stations.value_counts().reset_index()
station_usage.columns = ["station_name", "trip_count"]

top10 = station_usage.head(10)
print("\nTop 10 busiest stations:")
print(top10)
```

```python
import plotly.io as pio
pio.renderers.default = "jupyterlab" # if you’re in JupyterLab
```

```python
import plotly.express as px

# Combine start + end, count trips
all_stations = pd.concat([df['start_station_name'], df['end_station_name']])
station_usage = all_stations.value_counts().reset_index()
station_usage.columns = ["station_name", "trip_count"]

# Top 10 busiest stations
top10 = station_usage.head(10)

# Plot
fig_top10 = px.bar(
    top10,
    x="station_name",
    y="trip_count",
    title="Top 10 Busiest CitiBike Stations",
    text="trip_count",
    color="trip_count",
    color_continuous_scale="Blues"
)

fig_top10.update_layout(
    xaxis_title="Station Name",
    yaxis_title="Number of Trips",
    xaxis_tickangle=45,
    width=900,
    height=600
)

fig_top10.show()

# to add to streamlit
# st.plotly_chart(fig_top10, use_container_width=True)
```

```python
# Count trips starting at each station
start_counts = df['start_station_name'].value_counts().reset_index()
start_counts.columns = ['station_name', 'start_trips']

# Count trips ending at each station
end_counts = df['end_station_name'].value_counts().reset_index()
end_counts.columns = ['station_name', 'end_trips']

# Merge into one dataframe
station_counts = pd.merge(start_counts, end_counts, on='station_name', how='outer').fillna(0)

# Total trips = start + end
station_counts['total_trips'] = station_counts['start_trips'] + station_counts['end_trips']

# Top 10 stations by total trips
top10 = station_counts.nlargest(10, 'total_trips')
```

```python
import plotly.graph_objects as go

fig_top10_compare = go.Figure(data=[
    go.Bar(name='Start Trips', x=top10['station_name'], y=top10['start_trips']),
    go.Bar(name='End Trips', x=top10['station_name'], y=top10['end_trips'])
])

fig_top10_compare.update_layout(
    barmode='group',  # side-by-side bars
    title="Top 10 Busiest Stations: Start vs End Trips",
    xaxis_title="Station Name",
    yaxis_title="Number of Trips",
    xaxis_tickangle=45,
    width=950,
    height=600
)

fig_top10_compare.show()
```

```python
# for streamlit use
# st.plotly_chart(fig_top10_compare, use_container_width=True)
```

```python
fig_top10_stacked = go.Figure(data=[
    go.Bar(name='Start Trips', x=top10['station_name'], y=top10['start_trips']),
    go.Bar(name='End Trips', x=top10['station_name'], y=top10['end_trips'])
])

fig_top10_stacked.update_layout(
    barmode='stack',  # 👈 stacked instead of grouped
    title="Top 10 Busiest Stations: Start vs End Trips (Stacked)",
    xaxis_title="Station Name",
    yaxis_title="Number of Trips",
    xaxis_tickangle=45,
    width=950,
    height=600
)

fig_top10_stacked.show()
```

```python
top10['start_pct'] = (top10['start_trips'] / top10['total_trips'] * 100).round(1)
top10['end_pct'] = (top10['end_trips'] / top10['total_trips'] * 100).round(1)
```

```python
fig_top10_stacked_pct = go.Figure(data=[
    go.Bar(name='Start Trips', x=top10['station_name'], y=top10['start_trips'],
           text=top10['start_pct'].astype(str) + '%', textposition='inside'),
    go.Bar(name='End Trips', x=top10['station_name'], y=top10['end_trips'],
           text=top10['end_pct'].astype(str) + '%', textposition='inside')
])

fig_top10_stacked_pct.update_layout(
    barmode='stack',
    title="Top 10 Busiest Stations: Start vs End Trips (with %)",
    xaxis_title="Station Name",
    yaxis_title="Number of Trips",
    xaxis_tickangle=45,
    width=950,
    height=600
)

fig_top10_stacked_pct.show()
```

# Are existing stations evenly distributed?

```python
# Unique start stations
start_stations = df[['start_station_name','start_lat','start_lng']].dropna().drop_duplicates()
start_stations = start_stations.rename(columns={
    'start_station_name': 'station_name',
    'start_lat': 'lat',
    'start_lng': 'lon'
})

# Unique end stations
end_stations = df[['end_station_name','end_lat','end_lng']].dropna().drop_duplicates()
end_stations = end_stations.rename(columns={
    'end_station_name': 'station_name',
    'end_lat': 'lat',
    'end_lng': 'lon'
})

# Combine and deduplicate
stations = pd.concat([start_stations, end_stations], ignore_index=True).drop_duplicates(subset=["station_name"])

print("Unique stations:", stations.shape[0])
stations.head()

```

```python
import plotly.express as px

fig_stations = px.scatter_map(
    stations, lat="lat", lon="lon", hover_name="station_name",
    zoom=11, height=600
)
fig_stations.update_layout(mapbox_style="carto-darkmatter")
fig_stations.show()
```

```python
import pydeck as pdk

view_state = pdk.ViewState(latitude=40.75, longitude=-73.97, zoom=11)

layer = pdk.Layer(
    "HexagonLayer",
    data=stations,
    get_position=["lon", "lat"],
    radius=300,           # size of hex bins in meters
    elevation_scale=20,   # tower height
    elevation_range=[0, 1000],
    extruded=True,
    pickable=True,
)

r = pdk.Deck(layers=[layer], initial_view_state=view_state, map_style="mapbox://styles/mapbox/dark-v9")
r.show()
```

```python
import plotly.express as px

fig_density = px.density_mapbox(
    stations, 
    lat="lat", lon="lon", 
    radius=150,  # adjust radius for smoother/rougher look
    center={"lat": 40.75, "lon": -73.97}, 
    zoom=11,
    mapbox_style="carto-darkmatter",
    title="CitiBike Station Density"
)

fig_density.show()
```
